```python
import os
import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
import json
import h5py
from glob import glob
import numpy as np
import torch.utils.data as data
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# ==================== VQ-VAE Components ====================

class ResidualLayer(nn.Module):
    def __init__(self, in_dim, h_dim, res_h_dim):
        super(ResidualLayer, self).__init__()
        self.res_block = nn.Sequential(
            nn.ReLU(True),
            nn.Conv1d(in_dim, res_h_dim, kernel_size=3, stride=1, padding=1, bias=False),
            nn.ReLU(True),
            nn.Conv1d(res_h_dim, h_dim, kernel_size=1, stride=1, bias=False)
        )

    def forward(self, x):
        return x + self.res_block(x)

class ResidualStack(nn.Module):
    def __init__(self, in_dim, h_dim, res_h_dim, n_res_layers):
        super(ResidualStack, self).__init__()
        self.stack = nn.ModuleList([ResidualLayer(in_dim, h_dim, res_h_dim) for _ in range(n_res_layers)])

    def forward(self, x):
        for layer in self.stack:
            x = layer(x)
        return F.relu(x)

class PointCloudEncoder(nn.Module):
    def __init__(self, h_dim, n_res_layers, res_h_dim):
        super(PointCloudEncoder, self).__init__()
        self.conv_stack = nn.Sequential(
            nn.Conv1d(3, h_dim // 2, kernel_size=3, stride=1, padding=1),
            nn.ReLU(),
            nn.Conv1d(h_dim // 2, h_dim, kernel_size=3, stride=1, padding=1),
            nn.ReLU(),
            nn.Conv1d(h_dim, h_dim, kernel_size=3, stride=1, padding=1),
            ResidualStack(h_dim, h_dim, res_h_dim, n_res_layers)
        )

    def forward(self, x):
        # x shape: (batch, 3, num_points)
        return self.conv_stack(x)

class PointCloudDecoder(nn.Module):
    def __init__(self, embedding_dim, h_dim, n_res_layers, res_h_dim):
        super(PointCloudDecoder, self).__init__()
        self.inverse_conv_stack = nn.Sequential(
            nn.ConvTranspose1d(embedding_dim, h_dim, kernel_size=3, stride=1, padding=1),
            ResidualStack(h_dim, h_dim, res_h_dim, n_res_layers),
            nn.ConvTranspose1d(h_dim, h_dim // 2, kernel_size=3, stride=1, padding=1),
            nn.ReLU(),
            nn.ConvTranspose1d(h_dim // 2, 3, kernel_size=3, stride=1, padding=1)
        )

    def forward(self, x):
        return self.inverse_conv_stack(x)

class VectorQuantizer(nn.Module):
    def __init__(self, n_e, e_dim, beta):
        super(VectorQuantizer, self).__init__()
        self.n_e = n_e
        self.e_dim = e_dim
        self.beta = beta
        
        self.embedding = nn.Embedding(self.n_e, self.e_dim)
        self.embedding.weight.data.uniform_(-1.0 / self.n_e, 1.0 / self.n_e)

    def forward(self, z):
        # z shape: (batch, embedding_dim, num_points)
        z = z.permute(0, 2, 1).contiguous()  # (batch, num_points, embedding_dim)
        z_flattened = z.view(-1, self.e_dim)  # (batch*num_points, embedding_dim)
        
        # Compute distances
        d = torch.sum(z_flattened ** 2, dim=1, keepdim=True) + \
            torch.sum(self.embedding.weight**2, dim=1) - 2 * \
            torch.matmul(z_flattened, self.embedding.weight.t())
        
        # Find closest encodings
        min_encoding_indices = torch.argmin(d, dim=1).unsqueeze(1)
        min_encodings = torch.zeros(min_encoding_indices.shape[0], self.n_e).to(z.device)
        min_encodings.scatter_(1, min_encoding_indices, 1)
        
        # Get quantized latent vectors
        z_q = torch.matmul(min_encodings, self.embedding.weight).view(z.shape)
        
        # Compute loss
        loss = torch.mean((z_q.detach()-z)**2) + self.beta * torch.mean((z_q - z.detach()) ** 2)
        
        # Preserve gradients
        z_q = z + (z_q - z).detach()
        
        # Perplexity
        e_mean = torch.mean(min_encodings, dim=0)
        perplexity = torch.exp(-torch.sum(e_mean * torch.log(e_mean + 1e-10)))
        
        # Reshape back
        z_q = z_q.permute(0, 2, 1).contiguous()  # (batch, embedding_dim, num_points)
        
        return loss, z_q, perplexity, min_encodings, min_encoding_indices

class PointCloudVQVAE(nn.Module):
    def __init__(self, h_dim, res_h_dim, n_res_layers, n_embeddings, embedding_dim, beta):
        super(PointCloudVQVAE, self).__init__()
        
        self.encoder = PointCloudEncoder(h_dim, n_res_layers, res_h_dim)
        self.pre_quantization_conv = nn.Conv1d(h_dim, embedding_dim, kernel_size=1, stride=1)
        self.vector_quantization = VectorQuantizer(n_embeddings, embedding_dim, beta)
        self.decoder = PointCloudDecoder(embedding_dim, h_dim, n_res_layers, res_h_dim)

    def forward(self, x):
        # x shape: (batch, num_points, 3) -> (batch, 3, num_points)
        x = x.permute(0, 2, 1)
        
        # Encode
        z_e = self.encoder(x)
        z_e = self.pre_quantization_conv(z_e)
        
        # Quantize
        embedding_loss, z_q, perplexity, _, _ = self.vector_quantization(z_e)
        
        # Decode
        x_hat = self.decoder(z_q)
        
        # Back to original format: (batch, 3, num_points) -> (batch, num_points, 3)
        x_hat = x_hat.permute(0, 2, 1)
        
        return embedding_loss, x_hat, perplexity

# ==================== Dataset Class ====================

# ==================== Dataset Class ====================

class FilteredDataset(data.Dataset):
    def __init__(self, root_dir, num_points=256, split='train', max_samples=None):
        self.num_points = num_points
        self.filtered_dir = os.path.join(root_dir, "filtered_data")
        self.max_samples = max_samples if max_samples is not None else float('inf')  # Set to infinity if None
        
        # Load all data from filtered files
        self.data = []
        self.labels = []
        self.names = []
        
        # Class mapping
        self.class_names = ['bed', 'cabinet', 'table']
        self.class_to_label = {name: idx for idx, name in enumerate(self.class_names)}
        
        total_loaded = 0
        
        print(f"Looking for data in: {self.filtered_dir}")
        print(f"Max samples: {'All available' if max_samples is None else max_samples}")
        
        # Iterate through each class folder
        for class_name in self.class_names:
            if total_loaded >= self.max_samples:
                break
                
            class_dir = os.path.join(self.filtered_dir, class_name)
            
            # Check if class directory exists
            if not os.path.exists(class_dir):
                print(f"Warning: Directory {class_dir} does not exist, skipping {class_name}")
                continue
            
            # Load data for this class
            h5_file = os.path.join(class_dir, f"{class_name}.h5")
            id2name_file = os.path.join(class_dir, f"{class_name}_id2name.json")
            
            if not os.path.exists(h5_file):
                print(f"Warning: H5 file {h5_file} missing for {class_name}, skipping")
                continue
                
            if not os.path.exists(id2name_file):
                print(f"Warning: ID2name file {id2name_file} missing for {class_name}, skipping")
                continue
            
            print(f"Loading data from {h5_file}")
            
            # Load data
            try:
                with h5py.File(h5_file, "r") as f:
                    print(f"Available keys in H5 file: {list(f.keys())}")
                    class_data = f["data"][:]
                    print(f"Loaded {len(class_data)} samples from {class_name}")
                    
                    if "label" in f:
                        class_labels = f["label"][:]
                        print(f"Found labels in H5 file for {class_name}")
                    else:
                        # Create labels based on class index
                        class_labels = np.full(len(class_data), self.class_to_label[class_name])
                        print(f"Created labels for {class_name} with value {self.class_to_label[class_name]}")
            except Exception as e:
                print(f"Error loading H5 file for {class_name}: {e}")
                continue
            
            # Load names
            try:
                with open(id2name_file, "r") as f:
                    class_names_list = json.load(f)
                    print(f"Loaded names for {class_name}, type: {type(class_names_list)}")
            except Exception as e:
                print(f"Error loading names file for {class_name}: {e}")
                continue
            
            # Limit samples for this class
            if max_samples is None:
                samples_to_take = len(class_data)  # Take all samples
            else:
                samples_to_take = min(len(class_data), self.max_samples - total_loaded)
            
            # Add data
            self.data.append(class_data[:samples_to_take])
            
            # Add labels - ensure they are 1D
            current_labels = class_labels[:samples_to_take]
            if current_labels.ndim > 1:
                current_labels = current_labels.flatten()
            self.labels.append(current_labels)
            
            # Handle names (could be list or dict)
            if isinstance(class_names_list, list):
                self.names.extend(class_names_list[:samples_to_take])
            elif isinstance(class_names_list, dict):
                self.names.extend([class_names_list.get(str(i), f"{class_name}_{i}") for i in range(samples_to_take)])
            else:
                # Fallback: create names
                self.names.extend([f"{class_name}_{i}" for i in range(samples_to_take)])
            
            total_loaded += samples_to_take
            print(f"Added {samples_to_take} samples from {class_name}. Total: {total_loaded}")
        
        # Concatenate all data
        if self.data:
            self.data = np.concatenate(self.data, axis=0)
            self.labels = np.concatenate(self.labels, axis=0)
            
            # Ensure labels are 1D integers
            self.labels = self.labels.flatten().astype(int)
        else:
            print("No data loaded!")
            self.data = np.array([])
            self.labels = np.array([])
            self.names = []
        
        # No final truncation when max_samples is None
        if max_samples is not None and len(self.data) > max_samples:
            self.data = self.data[:max_samples]
            self.labels = self.labels[:max_samples]
            self.names = self.names[:max_samples]
        
        print(f"\n=== Final Dataset Info ===")
        print(f"Total loaded: {len(self.data)} samples")
        print(f"Data shape: {self.data.shape if len(self.data) > 0 else 'Empty'}")
        print(f"Labels shape: {self.labels.shape if len(self.labels) > 0 else 'Empty'}")
        print(f"Labels type: {type(self.labels)}")
        print(f"Number of names: {len(self.names)}")
        
        if len(self.labels) > 0:
            try:
                unique_labels = np.unique(self.labels)
                print(f"Unique labels: {unique_labels}")
                label_counts = np.bincount(self.labels)
                print(f"Labels distribution: {label_counts}")
                for i, count in enumerate(label_counts):
                    class_name = list(self.class_to_label.keys())[i] if i < len(self.class_to_label) else f"class_{i}"
                    print(f"  {class_name}: {count} samples")
            except Exception as e:
                print(f"Error computing label distribution: {e}")
                print(f"Labels sample: {self.labels[:10] if len(self.labels) > 0 else 'Empty'}")
        
        print(f"Class mapping: {self.class_to_label}")

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        if idx >= len(self.data):
            raise IndexError(f"Index {idx} out of range for dataset of size {len(self.data)}")
        
        points = self.data[idx]
        if len(points) > self.num_points:
            points = points[:self.num_points]  # Truncate if too many points
        elif len(points) < self.num_points:
            # Pad if too few points (repeat last point)
            padding = np.tile(points[-1:], (self.num_points - len(points), 1))
            points = np.concatenate([points, padding], axis=0)
        
        points = torch.from_numpy(points).float()
        label = torch.tensor(self.labels[idx]).long()
        name = self.names[idx] if idx < len(self.names) else f"sample_{idx}"
        
        return points, label, name

# ==================== Training Functions ====================

def train_vqvae(model, dataloader, optimizer, device, epoch):
    model.train()
    total_loss = 0
    total_recon_loss = 0
    total_embedding_loss = 0
    total_perplexity = 0
    
    for batch_idx, (data, labels, names) in enumerate(dataloader):
        data = data.to(device)
        optimizer.zero_grad()
        
        embedding_loss, x_hat, perplexity = model(data)
        
        # Reconstruction loss (MSE)
        recon_loss = F.mse_loss(x_hat, data)
        
        # Total loss
        loss = recon_loss + embedding_loss
        
        loss.backward()
        optimizer.step()
        
        total_loss += loss.item()
        total_recon_loss += recon_loss.item()
        total_embedding_loss += embedding_loss.item()
        total_perplexity += perplexity.item()
        
        if batch_idx % 5 == 0:  # Print more frequently for smaller dataset
            print(f'Epoch {epoch}, Batch {batch_idx}: '
                  f'Loss: {loss.item():.4f}, '
                  f'Recon: {recon_loss.item():.4f}, '
                  f'Embed: {embedding_loss.item():.4f}, '
                  f'Perplexity: {perplexity.item():.2f}')
    
    avg_loss = total_loss / len(dataloader)
    avg_recon = total_recon_loss / len(dataloader)
    avg_embed = total_embedding_loss / len(dataloader)
    avg_perplexity = total_perplexity / len(dataloader)
    
    return avg_loss, avg_recon, avg_embed, avg_perplexity

def visualize_reconstruction(model, dataloader, device, num_samples=3):
    model.eval()
    with torch.no_grad():
        data, labels, names = next(iter(dataloader))
        
        # Ensure we don't request more samples than available
        available_samples = data.shape[0]
        actual_samples = min(num_samples, available_samples)
        
        data = data[:actual_samples].to(device)
        
        _, x_hat, _ = model(data)
        
        # Convert to numpy
        original = data.cpu().numpy()
        reconstructed = x_hat.cpu().numpy()
        
        fig = plt.figure(figsize=(15, 5 * actual_samples))
        
        for i in range(actual_samples):
            # Original
            ax1 = fig.add_subplot(actual_samples, 2, 2*i + 1, projection='3d')
            points = original[i]
            ax1.scatter(points[:, 0], points[:, 1], points[:, 2], s=1, alpha=0.6)
            ax1.set_title(f'Original - {names[i]}')
            
            # Reconstructed
            ax2 = fig.add_subplot(actual_samples, 2, 2*i + 2, projection='3d')
            points_recon = reconstructed[i]
            ax2.scatter(points_recon[:, 0], points_recon[:, 1], points_recon[:, 2], s=1, alpha=0.6)
            ax2.set_title(f'Reconstructed - {names[i]}')
        
        plt.tight_layout()
        plt.show()

# ==================== Main Training Code ====================

if __name__ == "__main__":
    # Set device
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"Using device: {device}")
    
    # Hyperparameters
    h_dim = 128
    res_h_dim = 64
    n_res_layers = 3
    n_embeddings = 128
    embedding_dim = 32
    beta = 0.25
    learning_rate = 1e-3
    batch_size = 64  
    num_epochs = 200  
    num_points = 256
    
    # Create dataset and dataloader
    root_dir = "."  # Current directory since filtered_data is in the workspace root
    train_dataset = FilteredDataset(root_dir, num_points=num_points, split='train', max_samples=None)
    train_loader = data.DataLoader(train_dataset, batch_size=batch_size, shuffle=True, num_workers=0)  # Set to 0 to avoid Windows multiprocessing issues
    
    # Create model
    model = PointCloudVQVAE(h_dim, res_h_dim, n_res_layers, n_embeddings, embedding_dim, beta)
    model = model.to(device)
    
    # Optimizer
    optimizer = optim.Adam(model.parameters(), lr=learning_rate)
    
    # Training loop
    print("Starting training...")
    train_losses = []
    
    for epoch in range(num_epochs):
        avg_loss, avg_recon, avg_embed, avg_perplexity = train_vqvae(
            model, train_loader, optimizer, device, epoch
        )
        
        train_losses.append(avg_loss)
        
        print(f'Epoch {epoch}: Avg Loss: {avg_loss:.4f}, '
              f'Avg Recon: {avg_recon:.4f}, '
              f'Avg Embed: {avg_embed:.4f}, '
              f'Avg Perplexity: {avg_perplexity:.2f}')
        
        # Visualize reconstruction every 5 epochs
        if epoch % 5 == 0:
            print("Visualizing reconstruction...")
            visualize_reconstruction(model, train_loader, device)
        
        # Save model checkpoint
        if epoch % 10 == 0:
            torch.save({
                'epoch': epoch,
                'model_state_dict': model.state_dict(),
                'optimizer_state_dict': optimizer.state_dict(),
                'loss': avg_loss,
            }, f'vqvae_pointcloud_epoch_{epoch}.pth')
    
    # Plot training loss
    plt.figure(figsize=(10, 6))
    plt.plot(train_losses)
    plt.title('VQ-VAE Training Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.grid(True)
    plt.show()
    
    # Final visualization
    print("Final reconstruction visualization...")
    visualize_reconstruction(model, train_loader, device, num_samples=3)
    
    # Save final model
    torch.save(model.state_dict(), 'vqvae_pointcloud_final.pth')
    print("Training completed!")
```

```python
# ==================== Inference Functions ====================

class VQVAEInference:
    """Class để thực hiện inference riêng biệt cho VQ-VAE"""
    
    def __init__(self, model, device):
        self.model = model
        self.device = device
        self.model.eval()
    
    def pointcloud_to_codebook(self, pointcloud):
        """
        Chuyển từ 3D point cloud thành codebook indices
        
        Args:
            pointcloud: tensor shape (batch, num_points, 3) hoặc (num_points, 3)
            
        Returns:
            codebook_indices: tensor shape (batch, num_points) - indices của codebook
            quantized_features: tensor shape (batch, embedding_dim, num_points) - features đã quantized
        """
        with torch.no_grad():
            # Đảm bảo input có đúng shape
            if pointcloud.dim() == 2:
                pointcloud = pointcloud.unsqueeze(0)  # Thêm batch dimension
            
            pointcloud = pointcloud.to(self.device)
            
            # Encode: pointcloud -> features
            x = pointcloud.permute(0, 2, 1)  # (batch, 3, num_points)
            z_e = self.model.encoder(x)
            z_e = self.model.pre_quantization_conv(z_e)
            
            # Vector Quantization: features -> codebook indices
            embedding_loss, z_q, perplexity, min_encodings, min_encoding_indices = \
                self.model.vector_quantization(z_e)
            
            # Reshape indices để dễ sử dụng
            batch_size, embedding_dim, num_points = z_q.shape
            codebook_indices = min_encoding_indices.squeeze().view(batch_size, num_points)
            
            return codebook_indices, z_q, {
                'perplexity': perplexity.item(),
                'num_unique_codes': len(torch.unique(codebook_indices)),
                'embedding_loss': embedding_loss.item()
            }
    
    def codebook_to_pointcloud(self, codebook_indices, embedding_dim=None, num_points=None):
        """
        Chuyển từ codebook indices thành 3D point cloud
        
        Args:
            codebook_indices: tensor shape (batch, num_points) hoặc (num_points,) - indices của codebook
            embedding_dim: int - dimension của embedding (nếu None sẽ lấy từ model)
            num_points: int - số điểm (nếu None sẽ lấy từ input shape)
            
        Returns:
            pointcloud: tensor shape (batch, num_points, 3) - reconstructed point cloud
        """
        with torch.no_grad():
            # Đảm bảo input có đúng shape
            if codebook_indices.dim() == 1:
                codebook_indices = codebook_indices.unsqueeze(0)  # Thêm batch dimension
            
            codebook_indices = codebook_indices.to(self.device)
            batch_size, num_points = codebook_indices.shape
            
            if embedding_dim is None:
                embedding_dim = self.model.vector_quantization.e_dim
            
            # Tạo quantized features từ codebook indices
            z_q = self.model.vector_quantization.embedding(codebook_indices)  # (batch, num_points, embedding_dim)
            z_q = z_q.permute(0, 2, 1)  # (batch, embedding_dim, num_points)
            
            # Decode: quantized features -> pointcloud
            x_hat = self.model.decoder(z_q)
            x_hat = x_hat.permute(0, 2, 1)  # (batch, num_points, 3)
            
            return x_hat
    
    def encode_pointcloud(self, pointcloud):
        """Chỉ encode pointcloud thành features (không quantize)"""
        with torch.no_grad():
            if pointcloud.dim() == 2:
                pointcloud = pointcloud.unsqueeze(0)
            
            pointcloud = pointcloud.to(self.device)
            x = pointcloud.permute(0, 2, 1)
            z_e = self.model.encoder(x)
            z_e = self.model.pre_quantization_conv(z_e)
            
            return z_e
    
    def decode_features(self, features):
        """Chỉ decode features thành pointcloud (không quantize)"""
        with torch.no_grad():
            features = features.to(self.device)
            x_hat = self.model.decoder(features)
            x_hat = x_hat.permute(0, 2, 1)
            
            return x_hat

# ==================== Utility Functions ====================

def save_codebook_indices(codebook_indices, filename):
    """Lưu codebook indices ra file"""
    if isinstance(codebook_indices, torch.Tensor):
        codebook_indices = codebook_indices.cpu().numpy()
    
    np.save(filename, codebook_indices)
    print(f"Codebook indices saved to {filename}")

def load_codebook_indices(filename):
    """Load codebook indices từ file"""
    indices = np.load(filename)
    return torch.from_numpy(indices)

def visualize_codebook_usage(codebook_indices, codebook_size=128):
    """Visualize việc sử dụng codebook"""
    if isinstance(codebook_indices, torch.Tensor):
        codebook_indices = codebook_indices.cpu().numpy()
    
    # Flatten if needed
    flat_indices = codebook_indices.flatten()
    
    # Count usage
    usage_counts = np.bincount(flat_indices, minlength=codebook_size)
    used_codes = np.sum(usage_counts > 0)
    
    plt.figure(figsize=(12, 4))
    
    # Usage histogram
    plt.subplot(1, 2, 1)
    plt.hist(flat_indices, bins=min(50, codebook_size//10), alpha=0.7)
    plt.title(f'Codebook Usage Distribution\nUsed: {used_codes}/{codebook_size} codes')
    plt.xlabel('Codebook Index')
    plt.ylabel('Frequency')
    
    # Usage bar plot (top 20 most used)
    plt.subplot(1, 2, 2)
    top_indices = np.argsort(usage_counts)[-20:]
    plt.bar(range(20), usage_counts[top_indices])
    plt.title('Top 20 Most Used Codes')
    plt.xlabel('Rank')
    plt.ylabel('Usage Count')
    
    plt.tight_layout()
    plt.show()
    
    return {
        'total_codes': codebook_size,
        'used_codes': used_codes,
        'usage_percentage': used_codes / codebook_size * 100,
        'usage_counts': usage_counts
    }
```

```python
# ==================== Enhanced Inference để output discrete codes với visualization ====================

# 1. Load model và tạo inference object
device = torch.device("cpu")
model = PointCloudVQVAE(128, 64, 3, 128, 32, 0.25)  # Fixed: codebook_size=128, embedding_dim

# Load model
try:
    checkpoint = torch.load('vqvae_pointcloud_epoch_50.pth', map_location=device)
    if isinstance(checkpoint, dict) and 'model_state_dict' in checkpoint:
        model.load_state_dict(checkpoint['model_state_dict'])
        print(f"✅ Model loaded successfully from epoch {checkpoint.get('epoch', 'unknown')}")
    else:
        model.load_state_dict(checkpoint)
        print("✅ Model loaded successfully")
except:
    print("⚠️ Using random weights for demo...")

model = model.to(device)
inference = VQVAEInference(model, device)

# 2. Load dữ liệu từ dataset
print("Loading test data...")
test_dataset = FilteredDataset(".", num_points=256, split='train', max_samples=5)
test_loader = data.DataLoader(test_dataset, batch_size=1, shuffle=False, num_workers=0)

# 3. Process và output discrete codes với enhanced visualization
all_discrete_codes = []
all_metadata = []

for i, (pointcloud, label, name) in enumerate(test_loader):
    print(f"\n--- Processing sample {i+1}: {name[0]} ---")
    
    # Point Cloud -> Discrete Codes
    codebook_indices, quantized_features, stats = inference.pointcloud_to_codebook(pointcloud)
    
    # ✨ DISCRETE CODES
    discrete_codes = codebook_indices[0]  # Shape: (256,) - Fixed: now using 256 points
    
    print(f"✅ Discrete Codes Shape: {discrete_codes.shape}")
    print(f"   Codes range: {discrete_codes.min().item()} to {discrete_codes.max().item()}")
    print(f"   Sample codes: {discrete_codes[:20].tolist()}")
    print(f"   Unique codes used: {stats['num_unique_codes']}/128")  # Fixed: codebook size 128
    print(f"   Perplexity: {stats['perplexity']:.2f}")
    
    # Reconstruct từ discrete codes
    reconstructed_pc = inference.codebook_to_pointcloud(codebook_indices)
    mse_error = F.mse_loss(reconstructed_pc, pointcloud).item()
    
    # Lưu discrete codes
    discrete_codes_np = discrete_codes.cpu().numpy()
    np.save(f"discrete_codes_{name[0]}_sample_{i}.npy", discrete_codes_np)
    
    with open(f"discrete_codes_{name[0]}_sample_{i}.txt", "w") as f:
        codes_str = " ".join(map(str, discrete_codes_np.tolist()))
        f.write(codes_str)
    
    # JSON với metadata
    codes_data = {
        "sample_name": name[0],
        "sample_index": i,
        "label": label[0].item(),
        "discrete_codes": discrete_codes_np.tolist(),
        "num_points": len(discrete_codes_np),
        "unique_codes_used": stats['num_unique_codes'],
        "perplexity": stats['perplexity'],
        "codebook_size": 128,  # Fixed: codebook size 128
        "embedding_dim": 32,  # Added: embedding dimension
        "mse_error": mse_error
    }
    
    with open(f"discrete_codes_{name[0]}_sample_{i}.json", "w") as f:
        json.dump(codes_data, f, indent=2)
    
    # Thêm vào danh sách
    all_discrete_codes.append(discrete_codes_np)
    all_metadata.append({
        "name": name[0],
        "label": label[0].item(),
        "stats": stats,
        "mse_error": mse_error
    })
    
    # ==================== ENHANCED VISUALIZATION ====================
    
    # Figure 1: Comprehensive Analysis
    fig = plt.figure(figsize=(20, 12))
    
    # === ROW 1: 3D Point Clouds Comparison ===
    
    # Original Point Cloud (full view)
    ax1 = fig.add_subplot(3, 4, 1, projection='3d')
    points_orig = pointcloud[0].cpu().numpy()
    ax1.scatter(points_orig[:, 0], points_orig[:, 1], points_orig[:, 2], 
               s=0.5, alpha=0.8, c='blue', label='Original')
    ax1.set_title(f'Original Point Cloud\n{name[0]}', fontsize=12, fontweight='bold')
    ax1.set_xlabel('X')
    ax1.set_ylabel('Y')
    ax1.set_zlabel('Z')
    
    # Reconstructed Point Cloud (full view)
    ax2 = fig.add_subplot(3, 4, 2, projection='3d')
    points_recon = reconstructed_pc[0].cpu().numpy()
    ax2.scatter(points_recon[:, 0], points_recon[:, 1], points_recon[:, 2], 
               s=0.5, alpha=0.8, c='red', label='Reconstructed')
    ax2.set_title(f'Reconstructed Point Cloud\nMSE: {mse_error:.6f}', fontsize=12, fontweight='bold')
    ax2.set_xlabel('X')
    ax2.set_ylabel('Y')
    ax2.set_zlabel('Z')
    
    # Side-by-side overlay comparison
    ax3 = fig.add_subplot(3, 4, 3, projection='3d')
    # Sample points for better visibility
    sample_idx = np.random.choice(len(points_orig), min(256, len(points_orig)), replace=False)  # Fixed: using 256 points
    ax3.scatter(points_orig[sample_idx, 0], points_orig[sample_idx, 1], points_orig[sample_idx, 2], 
               s=1, alpha=0.7, c='blue', label='Original')
    ax3.scatter(points_recon[sample_idx, 0], points_recon[sample_idx, 1], points_recon[sample_idx, 2], 
               s=1, alpha=0.7, c='red', label='Reconstructed')
    ax3.set_title('Overlay Comparison\n(256 points)', fontsize=12, fontweight='bold')  # Fixed label
    ax3.legend()
    ax3.set_xlabel('X')
    ax3.set_ylabel('Y')
    ax3.set_zlabel('Z')
    
    # Error visualization (distance between original and reconstructed points)
    ax4 = fig.add_subplot(3, 4, 4, projection='3d')
    point_errors = np.linalg.norm(points_orig - points_recon, axis=1)
    scatter = ax4.scatter(points_orig[::10, 0], points_orig[::10, 1], points_orig[::10, 2], 
                         c=point_errors[::10], s=1, alpha=0.8, cmap='viridis')
    ax4.set_title(f'Point-wise Error\nMax: {point_errors.max():.4f}', fontsize=12, fontweight='bold')
    plt.colorbar(scatter, ax=ax4, shrink=0.8)
    ax4.set_xlabel('X')
    ax4.set_ylabel('Y')
    ax4.set_zlabel('Z')
    
    # === ROW 2: Discrete Codes Analysis ===
    
    # Discrete codes sequence
    ax5 = fig.add_subplot(3, 4, 5)
    ax5.plot(discrete_codes_np[:256], linewidth=0.8)  # Fixed: plot all 256 codes
    ax5.set_title(f'Discrete Codes Sequence\n(All 256 points)', fontsize=12, fontweight='bold')
    ax5.set_xlabel('Point Index')
    ax5.set_ylabel('Code Index (0-127)')  # Fixed: codebook range 0-127
    ax5.grid(True, alpha=0.3)
    
    # Codes histogram
    ax6 = fig.add_subplot(3, 4, 6)
    ax6.hist(discrete_codes_np, bins=min(32, 128//4), alpha=0.7, edgecolor='black', color='green')  # Fixed: bins appropriate for 128 codes
    ax6.set_title(f'Codes Distribution\n{stats["num_unique_codes"]}/128 codes used',  # Fixed: codebook size 128
                  fontsize=12, fontweight='bold')
    ax6.set_xlabel('Code Index')
    ax6.set_ylabel('Frequency')
    ax6.grid(True, alpha=0.3)
    
    # Most frequent codes
    ax7 = fig.add_subplot(3, 4, 7)
    unique, counts = np.unique(discrete_codes_np, return_counts=True)
    top_10_idx = np.argsort(counts)[-10:]
    ax7.bar(range(10), counts[top_10_idx], color='orange')
    ax7.set_title('Top 10 Most Used Codes', fontsize=12, fontweight='bold')
    ax7.set_xlabel('Rank')
    ax7.set_ylabel('Usage Count')
    ax7.set_xticks(range(10))
    ax7.set_xticklabels([f'{unique[i]}' for i in top_10_idx], rotation=45)
    
    # Codebook utilization
    ax8 = fig.add_subplot(3, 4, 8)
    utilization = stats['num_unique_codes'] / 128 * 100  # Fixed: codebook size 128
    ax8.pie([utilization, 100-utilization], 
           labels=[f'Used\n{stats["num_unique_codes"]}', f'Unused\n{128-stats["num_unique_codes"]}'],  # Fixed: codebook size 128
           colors=['lightgreen', 'lightcoral'], autopct='%1.1f%%', startangle=90)
    ax8.set_title(f'Codebook Utilization\n{utilization:.1f}%', fontsize=12, fontweight='bold')
    
    # === ROW 3: Statistical Analysis ===
    
    # Reconstruction error distribution
    ax9 = fig.add_subplot(3, 4, 9)
    ax9.hist(point_errors, bins=50, alpha=0.7, color='purple', edgecolor='black')
    ax9.set_title(f'Point-wise Error Distribution\nMean: {point_errors.mean():.6f}', 
                  fontsize=12, fontweight='bold')
    ax9.set_xlabel('Error (L2 distance)')
    ax9.set_ylabel('Number of Points')
    ax9.grid(True, alpha=0.3)
    
    # Coordinate comparison
    ax10 = fig.add_subplot(3, 4, 10)
    coords = ['X', 'Y', 'Z']
    orig_std = np.std(points_orig, axis=0)
    recon_std = np.std(points_recon, axis=0)
    x_pos = np.arange(len(coords))
    ax10.bar(x_pos - 0.2, orig_std, 0.4, label='Original', alpha=0.7)
    ax10.bar(x_pos + 0.2, recon_std, 0.4, label='Reconstructed', alpha=0.7)
    ax10.set_title('Coordinate Standard Deviation', fontsize=12, fontweight='bold')
    ax10.set_xlabel('Coordinate')
    ax10.set_ylabel('Standard Deviation')
    ax10.set_xticks(x_pos)
    ax10.set_xticklabels(coords)
    ax10.legend()
    ax10.grid(True, alpha=0.3)
    
    # Perplexity and metrics summary
    ax11 = fig.add_subplot(3, 4, 11)
    metrics = ['MSE', 'Perplexity', 'Unique Codes', 'Utilization %']
    values = [mse_error, stats['perplexity'], stats['num_unique_codes'], utilization]
    colors = ['red', 'blue', 'green', 'orange']
    bars = ax11.bar(metrics, values, color=colors, alpha=0.7)
    ax11.set_title('Key Metrics Summary', fontsize=12, fontweight='bold')
    ax11.set_ylabel('Value')
    
    # Add value labels on bars
    for bar, value in zip(bars, values):
        height = bar.get_height()
        ax11.text(bar.get_x() + bar.get_width()/2., height + height*0.01,
                 f'{value:.3f}', ha='center', va='bottom', fontsize=10)
    
    # Sample info
    ax12 = fig.add_subplot(3, 4, 12)
    ax12.axis('off')
    info_text = f"""
Sample Information:
• Name: {name[0]}
• Label: {label[0].item()}
• Points: {len(points_orig)}
• MSE Error: {mse_error:.6f}
• Perplexity: {stats['perplexity']:.2f}
• Unique Codes: {stats['num_unique_codes']}/128
• Utilization: {utilization:.1f}%
• Codebook Size: 128
• Embedding Dim: 256

Files Saved:
• {name[0]}_sample_{i}.npy
• {name[0]}_sample_{i}.txt  
• {name[0]}_sample_{i}.json
    """
    ax12.text(0.1, 0.9, info_text, transform=ax12.transAxes, fontsize=10,
             verticalalignment='top', bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.8))
    
    plt.tight_layout()
    plt.show()
    
    # ==================== SEPARATE 3D FIGURE FOR BETTER VIEW ====================
    
    # Figure 2: Large 3D comparison
    fig2 = plt.figure(figsize=(16, 8))
    
    # Original (large view)
    ax_orig = fig2.add_subplot(1, 2, 1, projection='3d')
    ax_orig.scatter(points_orig[:, 0], points_orig[:, 1], points_orig[:, 2], 
                   s=0.3, alpha=0.8, c='blue')
    ax_orig.set_title(f'Original Point Cloud - {name[0]}', fontsize=16, fontweight='bold')
    ax_orig.set_xlabel('X', fontsize=12)
    ax_orig.set_ylabel('Y', fontsize=12)
    ax_orig.set_zlabel('Z', fontsize=12)
    
    # Reconstructed (large view)
    ax_recon = fig2.add_subplot(1, 2, 2, projection='3d')
    ax_recon.scatter(points_recon[:, 0], points_recon[:, 1], points_recon[:, 2], 
                    s=0.3, alpha=0.8, c='red')
    ax_recon.set_title(f'Reconstructed Point Cloud\nMSE: {mse_error:.6f}', 
                      fontsize=16, fontweight='bold')
    ax_recon.set_xlabel('X', fontsize=12)
    ax_recon.set_ylabel('Y', fontsize=12)
    ax_recon.set_zlabel('Z', fontsize=12)
    
    plt.tight_layout()
    plt.show()
    
    if i >= 2:  # Process 3 samples
        break

# 6. Lưu tất cả discrete codes
print(f"\n=== Saving All Discrete Codes ===")
all_codes_data = {
    "total_samples": len(all_discrete_codes),
    "codebook_size": 128,  # Fixed: codebook size 128
    "embedding_dim": 32,  # Added: embedding dimension
    "points_per_sample": 256,
    "samples": []
}

for i, (codes, metadata) in enumerate(zip(all_discrete_codes, all_metadata)):
    all_codes_data["samples"].append({
        "index": i,
        "name": metadata["name"],
        "label": metadata["label"],
        "discrete_codes": codes.tolist(),
        "stats": metadata["stats"],
        "mse_error": metadata["mse_error"]
    })

# Lưu files
with open("all_discrete_codes.json", "w") as f:
    json.dump(all_codes_data, f, indent=2)

all_codes_array = np.array(all_discrete_codes)
np.save("all_discrete_codes.npy", all_codes_array)

print(f"✅ Saved {len(all_discrete_codes)} discrete code sequences")
print(f"   Shape: {all_codes_array.shape}")
print(f"   Files saved:")
print(f"   - all_discrete_codes.json (with metadata)")
print(f"   - all_discrete_codes.npy (numpy array)")
print(f"   - individual files for each sample")

# Summary statistics
print(f"\n=== Summary Statistics ===")
mse_errors = [m['mse_error'] for m in all_metadata]
perplexities = [m['stats']['perplexity'] for m in all_metadata]
unique_codes = [m['stats']['num_unique_codes'] for m in all_metadata]

print(f"Average MSE Error: {np.mean(mse_errors):.6f} ± {np.std(mse_errors):.6f}")
print(f"Average Perplexity: {np.mean(perplexities):.2f} ± {np.std(perplexities):.2f}")
print(f"Average Unique Codes: {np.mean(unique_codes):.1f} ± {np.std(unique_codes):.1f}")
print(f"Average Utilization: {np.mean(unique_codes)/128*100:.1f}%")  # Fixed: codebook size 128

print("\n✅ Enhanced discrete codes extraction completed!")
```
